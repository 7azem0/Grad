```python
import fastf1
import pandas as pd

races_to_inspect = [
    (2023, "Bahrain"),
    (2023, "Monaco"),
    (2023, "Silverstone"),
]

race_frames = []

for year, event in races_to_inspect:
    session = fastf1.get_session(year, event, "R")
    session.load()

    laps = session.laps.copy()
    laps["Year"] = year
    laps["Event"] = event

    race_frames.append(laps)

    print(f"\n{year} {event}")
    print("Rows and columns:", laps.shape)
    print("Available columns:")
    print(laps.columns.tolist())

audit_data = pd.concat(race_frames, ignore_index=True)

print("\nCombined shape:", audit_data.shape)
```

```python
required_columns = [
    "Driver",
    "LapTime",
    "Compound",
    "TyreLife",
    "LapNumber",
]

missing_columns = [
    column
    for column in required_columns
    if column not in audit_data.columns
]

print("Missing required columns:", missing_columns)
```

```python

# Make a separate copy so we don't change the original data
quality_data = audit_data.copy()

# Convert lap duration to seconds
quality_data["LapTimeSeconds"] = (
    quality_data["LapTime"].dt.total_seconds()
)

# 1. Inspect data quality for each race
for (year, event), race in quality_data.groupby(
    ["Year", "Event"]
):
    print(f"\n{'=' * 45}")
    print(f"{year} {event}")

    print("\nTotal rows:", len(race))
    print("Missing lap times:", race["LapTime"].isna().sum())

    print("\nAccuracy flags:")
    print(race["IsAccurate"].value_counts(dropna=False))

    print("\nDeleted flags:")
    print(race["Deleted"].value_counts(dropna=False))

    print("\nRows with pit entry time:", race["PitInTime"].notna().sum())
    print("Rows with pit exit time:", race["PitOutTime"].notna().sum())

    print("\nTrack status values:")
    print(race["TrackStatus"].value_counts(dropna=False))
```

```python

import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python

# Keep only the columns required for this experiment
model_data = candidate[
    ["Event", "Driver", "Compound", "TyreLife", "LapTimeSeconds"]
].copy()

# Train on two races; hold out the third race entirely
train_data = model_data[
    model_data["Event"].isin(["Bahrain", "Monaco"])
].copy()

test_data = model_data[
    model_data["Event"].eq("Silverstone")
].copy()

# Define X (inputs) and y (target)
features = ["Driver", "Compound", "TyreLife"]

X_train = train_data[features]
y_train = train_data["LapTimeSeconds"]

X_test = test_data[features]
y_test = test_data["LapTimeSeconds"]

print("Training laps:", len(X_train))
print("Testing laps:", len(X_test))
print("Training races:", train_data["Event"].unique())
print("Testing race:", test_data["Event"].unique())
```

```python

# Identify numeric and categorical inputs
categorical_features = ["Driver", "Compound"]
numeric_features = ["TyreLife"]

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features,
        ),
        ("numeric", "passthrough", numeric_features),
    ]
)

# Combine preprocessing and regression into one pipeline
model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regression", LinearRegression()),
    ]
)

# Learn from training data only
model.fit(X_train, y_train)

print("Model training complete!")
```

```python

# Predictions from linear regression
predictions = model.predict(X_test)

# Simple benchmark: always predict the training median
baseline = DummyRegressor(strategy="median")
baseline.fit(X_train, y_train)
baseline_predictions = baseline.predict(X_test)

def evaluate_predictions(actual, predicted):
    return {
        "MAE (seconds)": mean_absolute_error(actual, predicted),
        "RMSE (seconds)": np.sqrt(mean_squared_error(actual, predicted)),
        "R2": r2_score(actual, predicted),
    }

results = pd.DataFrame(
    {
        "Linear Regression": evaluate_predictions(
            y_test, predictions
        ),
        "Median Baseline": evaluate_predictions(
            y_test, baseline_predictions
        ),
    }
).T

print(results.round(3).to_string())
```

```python

# 1. Check the actual target distributions
print("TRAINING TARGET:")
print(y_train.describe())

print("\nTEST TARGET:")
print(y_test.describe())

# 2. Check the value predicted by the median baseline
print("\nBaseline constant prediction:")
print(baseline.constant_)

print("\nBaseline prediction distribution:")
print(pd.Series(baseline_predictions).describe())

# 3. Compare prediction lengths and errors
print("\nNumber of test targets:", len(y_test))
print("Number of regression predictions:", len(predictions))
print("Number of baseline predictions:", len(baseline_predictions))

print("\nManually calculated baseline MAE:")
print(np.mean(np.abs(y_test.to_numpy() - baseline_predictions)))

print("\nMedian of training targets:", y_train.median())
print("Median of test targets:", y_test.median())
```

```python

# Compare actual lap times with regression predictions
prediction_check = pd.DataFrame({
    "Actual": y_test.to_numpy(),
    "Predicted": predictions,
})

prediction_check["Error"] = (
    prediction_check["Actual"] - prediction_check["Predicted"]
)

print("Prediction summary:")
print(prediction_check.describe().round(2))

print("\nFirst 10 predictions:")
print(prediction_check.head(10).round(2).to_string(index=False))

print("\nLargest absolute errors:")
print(
    prediction_check.assign(
        AbsoluteError=prediction_check["Error"].abs()
    )
    .nlargest(10, "AbsoluteError")
    .round(2)
    .to_string(index=False)
)
```

```python

# Inspect the test predictions by lap-time range
prediction_check = pd.DataFrame({
    "Event": test_data["Event"].to_numpy(),
    "Actual": y_test.to_numpy(),
    "Predicted": predictions,
})

prediction_check["AbsoluteError"] = (
    prediction_check["Actual"] - prediction_check["Predicted"]
).abs()

print(
    prediction_check.groupby("Event").agg(
        Laps=("Actual", "count"),
        ActualMean=("Actual", "mean"),
        PredictedMean=("Predicted", "mean"),
        MAE=("AbsoluteError", "mean"),
    ).round(3).to_string()
)

print("\nPrediction error by actual lap-time range:")
prediction_check["ActualRange"] = pd.cut(
    prediction_check["Actual"],
    bins=[0, 92, 94, 97, float("inf")],
    labels=["Below 92 s", "92–94 s", "94–97 s", "Above 97 s"],
    right=False,
)

print(
    prediction_check.groupby("ActualRange", observed=False).agg(
        Laps=("Actual", "count"),
        ActualMean=("Actual", "mean"),
        PredictedMean=("Predicted", "mean"),
        MAE=("AbsoluteError", "mean"),
    ).round(3).to_string()
)
```

```python

# Select Bahrain only
bahrain_data = candidate.loc[
    candidate["Event"].eq("Bahrain")
].copy()

# Sort laps chronologically within each driver
bahrain_data = bahrain_data.sort_values(
    ["Driver", "LapNumber"]
).reset_index(drop=True)

print("Bahrain candidate laps:", len(bahrain_data))
print("Drivers:", bahrain_data["Driver"].nunique())

print("\nLaps per driver:")
print(
    bahrain_data.groupby("Driver")["LapNumber"]
    .agg(["min", "max", "count"])
    .to_string()
)
```

```python

# Each driver's earlier laps go into training;
# their later laps go into testing.
bahrain_data["LapPosition"] = (
    bahrain_data.groupby("Driver").cumcount()
)

bahrain_data["DriverLapCount"] = (
    bahrain_data.groupby("Driver")["LapNumber"]
    .transform("size")
)

bahrain_data["TrainCutoff"] = (
    bahrain_data["DriverLapCount"] * 0.70
).astype(int)

train_bahrain = bahrain_data.loc[
    bahrain_data["LapPosition"] < bahrain_data["TrainCutoff"]
].copy()

test_bahrain = bahrain_data.loc[
    bahrain_data["LapPosition"] >= bahrain_data["TrainCutoff"]
].copy()

print("Training laps:", len(train_bahrain))
print("Testing laps:", len(test_bahrain))

print("\nTraining lap ranges:")
print(
    train_bahrain.groupby("Driver")["LapNumber"]
    .agg(["min", "max"])
    .to_string()
)

print("\nTesting lap ranges:")
print(
    test_bahrain.groupby("Driver")["LapNumber"]
    .agg(["min", "max"])
    .to_string()
)
```

```python

features_v2 = ["Driver", "Compound", "Event", "TyreLife"]

X_train_v2 = train_bahrain[features_v2]
y_train_v2 = train_bahrain["LapTimeSeconds"]

X_test_v2 = test_bahrain[features_v2]
y_test_v2 = test_bahrain["LapTimeSeconds"]

categorical_v2 = ["Driver", "Compound", "Event"]
numeric_v2 = ["TyreLife"]

preprocessor_v2 = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_v2,
        ),
        ("numeric", "passthrough", numeric_v2),
    ]
)

model_v2 = Pipeline(
    steps=[
        ("preprocessor", preprocessor_v2),
        ("regression", LinearRegression()),
    ]
)

model_v2.fit(X_train_v2, y_train_v2)

predictions_v2 = model_v2.predict(X_test_v2)

print("Experiment 2 training complete!")
```

```python

baseline_v2 = DummyRegressor(strategy="median")
baseline_v2.fit(X_train_v2, y_train_v2)

baseline_predictions_v2 = baseline_v2.predict(X_test_v2)

results_v2 = pd.DataFrame({
    "Linear Regression": {
        "MAE": mean_absolute_error(y_test_v2, predictions_v2),
        "RMSE": np.sqrt(
            mean_squared_error(y_test_v2, predictions_v2)
        ),
        "R2": r2_score(y_test_v2, predictions_v2),
    },
    "Median Baseline": {
        "MAE": mean_absolute_error(
            y_test_v2, baseline_predictions_v2
        ),
        "RMSE": np.sqrt(
            mean_squared_error(y_test_v2, baseline_predictions_v2)
        ),
        "R2": r2_score(y_test_v2, baseline_predictions_v2),
    },
}).T

print(results_v2.round(3).to_string())
```

```python
# Sort laps in race order
bahrain_data = bahrain_data.sort_values(
    ["LapNumber", "Driver"]
).copy()

# Use one common cutoff for every driver
cutoff_lap = 40

train_data = bahrain_data[
    bahrain_data["LapNumber"] <= cutoff_lap
].copy()

test_data = bahrain_data[
    bahrain_data["LapNumber"] > cutoff_lap
].copy()

print("Training laps:", len(train_data))
print("Testing laps:", len(test_data))

print("\nTraining lap range:")
print(train_data["LapNumber"].min(), "to", train_data["LapNumber"].max())

print("\nTesting lap range:")
print(test_data["LapNumber"].min(), "to", test_data["LapNumber"].max())
```

```python
features = ["Driver", "Compound", "TyreLife"]

X_train = train_data[features]
y_train = train_data["LapTimeSeconds"]

X_test = test_data[features]
y_test = test_data["LapTimeSeconds"]

# Train linear regression
model.fit(X_train, y_train)
predictions = model.predict(X_test)

# Train the baseline on exactly the same training data
baseline.fit(X_train, y_train)
baseline_predictions = baseline.predict(X_test)

# Compare results
results = pd.DataFrame({
    "MAE": [
        mean_absolute_error(y_test, predictions),
        mean_absolute_error(y_test, baseline_predictions)
    ],
    "RMSE": [
        np.sqrt(mean_squared_error(y_test, predictions)),
        np.sqrt(mean_squared_error(y_test, baseline_predictions))
    ],
    "R2": [
        r2_score(y_test, predictions),
        r2_score(y_test, baseline_predictions)
    ]
}, index=["Linear Regression", "Median Baseline"])

print(results.round(3))
```
